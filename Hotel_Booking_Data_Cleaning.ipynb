#Filling null values in numerical columns with MEDIAN


```python
df['adr'] = df['adr'].astype(object)
df['stays_in_week_nights'] = df['stays_in_week_nights'].astype(object)
df['stays_in_weekend_nights'] = df['stays_in_weekend_nights'].astype(object)
```

```python
num_cols = df.select_dtypes(include = 'number').columns
num_cols
```

```python
for i in num_cols:
  df.loc[:,i] = df[i].fillna(df[i].median())
```

```python
df.isnull().sum()
```

#Filling null values in categorical columns with MODE.


```python
df['country'] = df['country'].fillna(df['country'].mode()[0])
```

```python
df.isnull().sum()
```

#Handling outliers for the target variables.

```python
df['adr'] = df['adr'].astype(float)
df['stays_in_week_nights'] = df['stays_in_week_nights'].astype(float)
df['stays_in_weekend_nights'] = df['stays_in_weekend_nights'].astype(float)
```

1.adr

```python
Q1 = df['adr'].quantile(0.25)
Q3 = df['adr'].quantile(0.75)
Q1,Q3
```

```python
IQR = Q3 - Q1
IQR
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
lower_limit, upper_limit
```

```python
df.shape
```

```python
df = df[(df['adr'] >= lower_limit) & (df['adr'] <= upper_limit)]
```
